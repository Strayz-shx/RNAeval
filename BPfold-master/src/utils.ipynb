```python
#!/usr/bin/env python3
import os
import glob

input_dir = "/home/bingxing2/home/scx7auf/Shihaoxuan/graduate/data_code/data_512/TestB_512/bpseq"      # 修改为你的文件夹路径
output_fasta = "/home/bingxing2/home/scx7auf/Shihaoxuan/graduate/data_code/data_512/TestB_512/testsetb.fasta"        # 输出的 fasta 文件名

with open(output_fasta, "w") as out:
    # 遍历所有 .bpseq 文件
    for filepath in sorted(glob.glob(os.path.join(input_dir, "*.bpseq"))):
        filename = os.path.basename(filepath)
        name = os.path.splitext(filename)[0]  # 去掉 .bpseq 作为标题
        
        # 读取并提取序列
        seq = []
        with open(filepath, "r") as f:
            for line in f:
                line = line.strip()
                if not line or line.startswith("#"):
                    continue
                parts = line.split()
                if len(parts) >= 2:
                    # bpseq 第二列是碱基 (A/U/G/C)
                    seq.append(parts[1])
        
        # 写入 fasta
        if seq:
            out.write(f">{name}\n")
            out.write("".join(seq) + "\n")

print(f"Done! Output: {output_fasta}")
```

```python
#!/usr/bin/env python3
import os
import csv
import math
import glob
import numpy as np
from scipy import stats

# ==================== 配置路径 ====================
PRED_DIR = "/home/bingxing2/home/scx7auf/Shihaoxuan/graduate/data_code/data_512/TestB_512/bpfold"      # BPfold 预测 bpseq 文件夹
TRUE_DIR = "/home/bingxing2/home/scx7auf/Shihaoxuan/graduate/data_code/data_512/TestB_512/bpseq"          # 真实结构 bpseq 文件夹
CI_CSV = "/home/bingxing2/home/scx7auf/Shihaoxuan/graduate/data_code/data_512/TestB_512/bpfold/ci_summary.csv"        # 包含 CI 的 CSV 文件

# ==================== 1. 读取 CI CSV ====================
def load_ci_csv(csv_path):
    """
    读取 CSV，返回 {seq_name: CI}。
    格式要求：第一列 = seq_name（与 bpseq 文件名对应，不含扩展名）
             第二列 = CI 值
    自动检测分隔符（逗号或制表符），跳过表头行。
    """
    ci_dict = {}
    with open(csv_path, 'r', newline='', encoding='utf-8') as f:
        sample = f.read(4096)
        f.seek(0)
        delimiter = '\t' if '\t' in sample else ','
        reader = csv.reader(f, delimiter=delimiter)

        for i, row in enumerate(reader):
            if not row or all(c == '' for c in row):
                continue
            # 跳过表头行（如果第一列包含非数字的表头关键词）
            if i == 0:
                first_col = row[0].strip().lower()
                if any(h in first_col for h in ['seq_name', 'name', 'id', 'filename', 'file']):
                    continue

            if len(row) >= 2:
                seq_name = row[0].strip()
                ci_str = row[1].strip()
                try:
                    ci_dict[seq_name] = float(ci_str)
                except ValueError:
                    continue
    return ci_dict


# ==================== 2. MCC 计算 ====================
def evaluate_similarity(pred_path, true_path):
    """计算两个 bpseq 文件的 Matthew's Correlation Coefficient (MCC)。"""
    nodes1, nodes2 = [], []
    with open(pred_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 3:
                nodes1.append(int(parts[2]))
    with open(true_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 3:
                nodes2.append(int(parts[2]))

    if len(nodes1) != len(nodes2):
        print(f"  [警告] 长度不匹配: {os.path.basename(pred_path)} ({len(nodes1)} vs {len(nodes2)})")
        return None

    tp = tn = fp = fn = 0
    for v1, v2 in zip(nodes1, nodes2):
        if v1 == v2 and v1 != 0:
            tp += 1
        elif v1 == v2 and v1 == 0:
            tn += 1
        elif v1 != 0 and v2 == 0:
            fp += 1
        elif v1 == 0 and v2 != 0:
            fn += 1
        else:
            # 两者都不为0但不相等（配对预测错误）
            fp += 1
            fn += 1

    denominator = math.sqrt((tp + fp) * (tp + fn) * (tn + fp) * (tn + fn))
    mcc = ((tp * tn) - (fp * fn)) / denominator if denominator != 0 else 0.0
    return mcc


# ==================== 3. 主分析流程 ====================
def analyze_ci_mcc(pred_dir, true_dir, ci_csv):
    ci_dict = load_ci_csv(ci_csv)
    print(f"[信息] CSV 中读取到 {len(ci_dict)} 个 CI 记录")
    print(f"[信息] 前10个 seq_name 示例: {list(ci_dict.keys())[:10]}\n")

    pred_files = sorted(glob.glob(os.path.join(pred_dir, "*.bpseq")))
    print(f"[信息] 预测文件夹中找到 {len(pred_files)} 个 .bpseq 文件")
    print(f"[信息] 前10个文件名示例: {[os.path.basename(f) for f in pred_files[:10]]}\n")

    ci_list = []
    mcc_norm_list = []
    skipped = []

    for pred_bpseq in pred_files:
        basename = os.path.basename(pred_bpseq)
        seq_name = os.path.splitext(basename)[0]  # 去掉 .bpseq 扩展名

        # 直接按文件名（不含扩展名）匹配 CI
        ci = ci_dict.get(seq_name)
        if ci is None:
            skipped.append(f"{seq_name}(CSV无匹配)")
            continue

        # 匹配真实文件（同名 .bpseq）
        true_bpseq = os.path.join(true_dir, basename)
        if not os.path.exists(true_bpseq):
            skipped.append(f"{seq_name}(无真实文件)")
            continue

        mcc = evaluate_similarity(pred_bpseq, true_bpseq)
        if mcc is None:
            skipped.append(f"{seq_name}(长度不匹配)")
            continue

        mcc_norm = (mcc + 1.0) / 2.0
        ci_list.append(ci)
        mcc_norm_list.append(mcc_norm)
        print(f"{seq_name:<20} CI={ci:.4f}  MCC={mcc:.4f}  MCC_norm={mcc_norm:.4f}")

    n = len(ci_list)
    print(f"\n{'='*60}")
    print(f"诊断统计")
    print(f"{'='*60}")
    print(f"  预测文件总数 : {len(pred_files)}")
    print(f"  成功匹配数   : {n}")
    print(f"  跳过/失败    : {len(skipped)}")
    if skipped:
        print(f"  跳过原因示例 : {', '.join(skipped[:5])}" + (" ..." if len(skipped) > 5 else ""))
    print(f"{'='*60}")

    if n < 2:
        print(f"[错误] 有效样本仅 {n} 个，无法计算相关性")
        return

    ci_arr = np.array(ci_list)
    mcc_arr = np.array(mcc_norm_list)

    pearson_r, pearson_p = stats.pearsonr(ci_arr, mcc_arr)
    r_squared = pearson_r ** 2
    mae = np.mean(np.abs(ci_arr - mcc_arr))

    print(f"\n{'='*60}")
    print(f"CI 与 归一化 MCC 相关性分析")
    print(f"{'='*60}")
    print(f"Pearson r  : {pearson_r:.4f}")
    print(f"p-value    : {pearson_p:.2e}")
    print(f"R² (r²)    : {r_squared:.4f}")
    print(f"MAE        : {mae:.4f}")
    print(f"{'='*60}")


analyze_ci_mcc(PRED_DIR, TRUE_DIR, CI_CSV)
```
